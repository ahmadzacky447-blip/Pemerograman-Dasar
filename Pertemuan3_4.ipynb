```python
#MENGHITUNG NILAI AKHIR MAHASISWA
tugas = 80
uts = 70
uas = 90

nilai_akhir = (0.3 * tugas) + (0.3 * uts) + (0.4 * uas)
print("Nilai akhir =", nilai_akhir)

# Modifikasi 2 : nilai_mahasiswa dihitung dari 5 variabel nilai (input keyboard)
tugas = float(input("Masukkan nilai tugas      : "))
kuis = float(input("Masukkan nilai kuis       : "))
praktikum = float(input("Masukkan nilai praktikum  : "))
uts = float(input("Masukkan nilai UTS        : "))
uas = float(input("Masukkan nilai UAS        : "))
standar_kelulusan = float(input("Masukkan standar kelulusan  : "))

# Bobot tiap komponen (total 100% = 1.0)
nilai_mahasiswa = (0.2 * tugas) + (0.1 * kuis) + (0.2 * praktikum) + (0.2 * uts) + (0.3 * uas)

apakah_lulus = nilai_mahasiswa >= standar_kelulusan
perlu_remedial = nilai_mahasiswa < standar_kelulusan

print(f"Nilai Mahasiswa: {nilai_mahasiswa:.2f}")
print(f"Standar Kelulusan: {standar_kelulusan}")
print(f"Apakah Mahasiswa Lulus? {apakah_lulus}")
print(f"Perlu Remedial? {perlu_remedial}")
```

```python
#SOAL 1 : BELANJA BUKU

jumlah_buku = 7
harga_buku = 18500
uang_budi = 150000
jumlah_orang = 4

total = jumlah_buku * harga_buku
diskon = total * 10 // 100
bayar = total - diskon
kembalian = uang_budi - bayar
per_orang = bayar // jumlah_orang
sisa = bayar % jumlah_orang
uang_cukup = uang_budi >= bayar

print("Total       :", total)
print("Diskon      :", diskon)
print("Bayar       :", bayar)
print("Kembalian   :", kembalian)
print("Per orang   :", per_orang)
print("Sisa        :", sisa)
print("Uang cukup? :", uang_cukup)
```

```python
#SOAL 2 : PLTS ATAP RUMAH

jumlah_panel = 12
daya_panel = 450        # Wp
jam_sinar = 5           # jam/hari
efisiensi = 0.8
kebutuhan = 25          # kWh/hari
tarif = 1444.70         # Rp/kWh
hari = 30

energi_harian = jumlah_panel * daya_panel * jam_sinar * efisiensi / 1000
energi_bulanan = energi_harian * hari
nilai_listrik = energi_bulanan * tarif
kekurangan = kebutuhan - energi_harian
mencukupi = energi_harian >= kebutuhan

print("Energi harian (kWh) :", round(energi_harian, 1))
print("Energi bulanan (kWh):", round(energi_bulanan, 1))
print("Nilai listrik (Rp)  :", round(nilai_listrik, 1))
print("Kekurangan (kWh)    :", round(kekurangan, 1))
print("Mencukupi?          :", mencukupi)
```

```python
#SOAL 3 : GENERATOR DIESEL & RUGI KABEL

daya = 200              # kW
jam = 8                 # jam/hari
konsumsi = 0.25         # liter/kWh
harga_solar = 13500     # Rp/liter
tangki = 1000           # liter
arus = 20               # A
hambatan = 0.5          # ohm

energi = daya * jam
solar = energi * konsumsi
biaya = solar * harga_solar
hari_tangki = tangki // solar
sisa_solar = tangki % solar
rugi_kabel = arus ** 2 * hambatan
rugi_tinggi = rugi_kabel > 150 and arus >= 20

print("Energi (kWh)     :", energi)
print("Solar (liter)    :", solar)
print("Biaya (Rp)       :", biaya)
print("Tangki cukup     :", hari_tangki, "hari")
print("Sisa solar (L)   :", sisa_solar)
print("Rugi kabel (W)   :", rugi_kabel)
print("Rugi tinggi?     :", rugi_tinggi)
```
