```python
# Soal 1. Daya Generator dan Efisiensi Turbin

# a. Menyimpan tegangan dan arus, lalu menghitung daya listrik
tegangan = 220          # tegangan generator (Volt)
arus = 15               # arus generator (Ampere)
daya = tegangan * arus  # P = V x I (Watt)
print(f"Daya listrik = {daya} W")

# b. Menghitung efisiensi turbin
energi_input = 500      # energi yang diterima turbin
energi_output = 400     # energi yang dihasilkan turbin
efisiensi = (energi_output / energi_input) * 100
print(f"Efisiensi = {efisiensi:.2f}%")
```

Penjelasan: tegangan dan arus saya simpan di variabel `tegangan` dan `arus`. Daya dihitung pakai rumus P = V x I, jadi 220 x 15 = 3300 W, lalu dicetak pakai f-string supaya nilainya langsung masuk ke kalimat. Untuk efisiensi, energi output dibagi energi input (400 / 500 = 0,8) lalu dikali 100 jadi 80%. Bagian `:.2f` dipakai supaya hasilnya selalu tampil dengan 2 angka di belakang koma, makanya keluarnya 80.00.

```python
# Soal 3. Data Suhu Boiler

suhu = [92.5, 88.7, 95.1, 102.3, 97.8]

# a. Menampilkan suhu pada titik ke-3 (indeks 2, karena indeks dimulai dari 0)
print(f"Suhu titik ke-3 = {suhu[2]} °C")

# b. Menampilkan tiga pembacaan pertama dengan slicing
print(f"Tiga pembacaan pertama = {suhu[:3]}")

# c. Menambahkan pembacaan baru, lalu menampilkan jumlah data
suhu.append(99.4)
print(f"Jumlah data setelah penambahan = {len(suhu)}")
```

Penjelasan: data suhu disimpan dalam satu list bernama `suhu`. Index list mulai dari 0, jadi titik ke-3 ada di `suhu[2]` dan hasilnya 95.1. Untuk tiga pembacaan pertama dipakai slicing `suhu[:3]`, yang mengambil index 0 sampai 2. Terakhir, `append(99.4)` menambah data baru di ujung list, lalu `len(suhu)` menghitung jumlah datanya, sekarang jadi 6.

```python
# Soal 5. Katup Pengaman

tekanan_uap = float(input("Masukkan tekanan uap (psi): "))
status_manual_override = True   # ubah ke True untuk menguji mode manual

if tekanan_uap > 150 and status_manual_override == False:
    print("Katup Terbuka")
elif status_manual_override == True:
    print("Katup Mode Manual")
else:
    print("Katup Tertutup")
```

Penjelasan: `input()` selalu menghasilkan teks, jadi saya ubah dulu ke float supaya bisa dibandingkan dengan angka 150. Di bagian `if`, ada dua syarat yang harus benar dua-duanya (pakai `and`), yaitu tekanan lebih dari 150 dan override False, baru katup terbuka. Kalau tidak, `elif` mengecek apakah override True, dan kalau iya langsung masuk mode manual berapa pun tekanannya. Selain dua kondisi itu, hasilnya Katup Tertutup. Perlu diperhatikan tanda `>` bukan `>=`, jadi tekanan 150 pas masih Tertutup.

```python
# Soal 7. Pemanas Mencapai Target

suhu_saat_ini = 60
target = 85

while suhu_saat_ini < target:
    suhu_saat_ini += 5                       # naikkan 5 derajat
    print(f"Suhu saat ini: {suhu_saat_ini} C")
    if suhu_saat_ini == 80:                  # peringatan tepat di 80
        print("WARNING: Mendekati target!")

print(f"Target {target} C tercapai.")
```

Penjelasan: `while` mengecek kondisi `suhu_saat_ini < target` di setiap awal putaran, jadi loop jalan terus sampai suhunya sudah 85. Di dalam loop suhu ditambah 5 pakai `+=`, baru setelah itu dicetak, makanya angka pertama yang muncul 65 dan bukan 60. Pesan WARNING muncul lewat `if` yang cek apakah suhu tepat 80. Pesan target tercapai ditulis di luar loop (tidak menjorok), jadi hanya tercetak sekali setelah loop selesai.

```python
# Soal 9. Pembacaan Vibrasi dengan Sensor Rusak

data = ["1.2", "0.9", "ERR", "3.1", "1.8"]
jumlah_tidak_valid = 0

for n, teks in enumerate(data, start=1):
    # a. Ubah teks menjadi float, tangani data yang gagal diubah
    try:
        nilai = float(teks)
    except ValueError:
        print(f"Unit {n}: data tidak valid")
        jumlah_tidak_valid += 1
        continue                      # lanjut ke unit berikutnya

    # b. Peringatan jika vibrasi melebihi 2.0
    if nilai > 2.0:
        print(f"Warning: Unit {n} vibrasi tinggi!")

        # c. Jika melebihi 3.0, hentikan pemeriksaan
        if nilai > 3.0:
            print("KRITIS: pemeriksaan dihentikan")
            break

# d. Tampilkan jumlah data tidak valid
print(f"Jumlah data tidak valid: {jumlah_tidak_valid}")
```

Penjelasan: `enumerate(data, start=1)` dipakai supaya nomor unit mulai dari 1. Di dalam loop, tiap data dicoba diubah ke float dengan `try`. Untuk "ERR" akan muncul ValueError, lalu ditangkap oleh `except`, dicetak pesan data tidak valid, penghitungnya ditambah, dan `continue` melompat ke unit berikutnya. Kalau berhasil, nilainya dicek: lebih dari 2.0 dicetak Warning, dan kalau lebih dari 3.0 dicetak KRITIS lalu `break` menghentikan loop, jadi Unit 5 tidak ikut diperiksa. Jumlah data tidak valid dicetak di luar loop setelah semuanya selesai.

```python
# CHALLENGE BONUS: Hitung Konsumsi dan Biaya Listrik Rumah
#
# Masalah:
# Di rumah ada beberapa alat listrik. Program ini menghitung berapa kWh
# yang terpakai per hari dan per bulan, alat mana yang paling boros,
# masuk kategori apa pemakaiannya, dan perkiraan biaya tagihannya.
#
# Alurnya:
# 1. Data alat (nama, daya dalam watt, jam pakai per hari) disimpan di list.
# 2. Pakai for untuk membaca tiap alat. Dayanya diubah ke float pakai
#    try-except supaya kalau ada data yang salah, program tidak error.
# 3. kWh per hari = daya x jam / 1000, hasilnya disimpan ke list.
# 4. Semua dijumlahkan, lalu if-elif-else menentukan kategorinya.
# 5. Biaya = kWh per bulan x tarif per kWh (tarifnya cuma perkiraan).


TARIF_PER_KWH = 1444.70   # asumsi tarif (Rp/kWh)
HARI_PER_BULAN = 30

# Data peralatan: (nama, daya dalam W sebagai teks, jam pakai per hari)
peralatan = [
    ("Lampu LED", "36", 10),
    ("Kulkas", "120", 24),
    ("AC", "400", 8),
    ("Televisi", "100", 5),
    ("Kipas Angin", "abc", 6),   # data rusak, sengaja untuk menguji try-except
    ("Pompa Air", "250", 2),
    ("Setrika", "300", 1),
]

daftar_nama = []          # nama peralatan dengan data valid
daftar_kwh = []           # konsumsi harian tiap peralatan (kWh)
jumlah_tidak_valid = 0

print("=== Konsumsi Listrik Harian ===")
for nama, daya_teks, jam in peralatan:
    try:
        daya = float(daya_teks)
    except ValueError:
        print(f"{nama}: data daya tidak valid ('{daya_teks}'), dilewati")
        jumlah_tidak_valid += 1
        continue

    kwh = daya * jam / 1000          # W x jam -> Wh, dibagi 1000 -> kWh
    daftar_nama.append(nama)
    daftar_kwh.append(kwh)
    print(f"{nama:<12}{daya:>5.0f} W x {jam:>2} jam = {kwh:.2f} kWh/hari")

# Perhitungan total
total_harian = sum(daftar_kwh)
total_bulanan = total_harian * HARI_PER_BULAN
biaya = total_bulanan * TARIF_PER_KWH

# Mencari peralatan paling boros
indeks_terbesar = daftar_kwh.index(max(daftar_kwh))
alat_terboros = daftar_nama[indeks_terbesar]

# Menentukan kategori pemakaian dengan if-elif-else
if total_bulanan <= 100:
    kategori = "Hemat"
    saran = "Konsumsi sudah efisien, pertahankan."
elif total_bulanan <= 300:
    kategori = "Sedang"
    saran = f"Kurangi jam pemakaian {alat_terboros} untuk menghemat."
else:
    kategori = "Boros"
    saran = f"Segera batasi pemakaian {alat_terboros} dan alat berdaya besar lain."

print("\n=== Ringkasan ===")
print(f"Total konsumsi per hari : {total_harian:.2f} kWh")
print(f"Estimasi per bulan      : {total_bulanan:.2f} kWh")
print(f"Alat paling boros       : {alat_terboros} ({max(daftar_kwh):.2f} kWh/hari)")
print(f"Kategori pemakaian      : {kategori}")
print(f"Estimasi biaya          : Rp {biaya:,.0f}")
print(f"Data tidak valid        : {jumlah_tidak_valid}")
print(f"Saran                   : {saran}")
```

Penjelasan: data peralatan disimpan di list `peralatan`, sedangkan hasil hitungannya masuk ke dua list lain, `daftar_nama` dan `daftar_kwh`. Tiap alat dibaca dengan `for`, dan daya yang berupa teks diubah ke float pakai try-except. Datanya sengaja saya isi "abc" di Kipas Angin untuk menguji bagian ini, hasilnya alat itu dilewati dan dihitung sebagai data tidak valid. Konsumsi per alat dihitung dengan daya x jam / 1000 supaya satuannya kWh, lalu dijumlahkan dengan `sum()`. Alat paling boros dicari dengan `max()` dan `.index()`. Kategori pemakaian ditentukan pakai if-elif-else (Hemat, Sedang, atau Boros) dan saran di akhir mengikuti kategorinya. Tarif listrik hanya perkiraan, bisa diganti di `TARIF_PER_KWH`.
